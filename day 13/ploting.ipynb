```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0,10,100)
y = np.sin(x)
plt.plot(x,y, label="sin(x)", color="blue", linestyle="--")
plt.title("Sine Function")
plt.xlabel("x")
plt.ylabel("sin(x)")
plt.grid(True)
plt.show()
```

```python
# create a scatter plot
x = np.random.rand(100)
y = np.random.rand(100)
plt.scatter(x,y, color = "red",alpha = 0.5)
plt.title("Random value")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True)
```

```python
# create a bar plot
category = ['A','B','C','D']
values = [10,20,15,35]

plt.bar(category,values, color=["orange",'blue','red','green'])
plt.title("bar plot")
plt.xlabel("category")
plt.ylabel("quantity")
```

```python
# create histogram plot
data = np.random.ranf(1000)

plt.hist(data,bins = 30 ,color = "blue",alpha = 0.8,edgecolor = "black")
plt.title("H1istogram")
plt.xlabel("Bins")
plt.ylabel("Frequency")
```

```python
# create pie chart
label = ["python","java","c++","javascript","database"]
sizes = [42,28,15,8,7]
plt.pie(sizes,labels = label,autopct="%1.1f%%",startangle=90)
plt.title("Programming language")
```
