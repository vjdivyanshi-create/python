```python
# 4) Write a Python program to calculate the total, highest value, and lowest value from a tuple of integers without using the built-in sum(), max(), and min() functions.

nums = (12, 45, 7, 23, 89, 15)
total = 0
highest = nums[0]
lowest = nums[0]
for num  in nums:
    total = total + num

    if num> highest:
        highest  = num

    if num< lowest:
        lowest = num

print("total:", total)
print("highest:",highest)
print("lowest:",lowest)
```
