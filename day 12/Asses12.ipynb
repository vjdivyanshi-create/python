```python
import numpy as np
```

```python
# 1 covert 1d into 2d array
a = np.array([1, 2, 3, 4, 5, 6])
a2 = a.reshape(2, 3)
print(a2)
```

```python
# 2 print array attributes 
a = np.array([[1, 2, 3], [4, 5, 6]])
print("Shape:", a.shape)
print("Data type:", a.dtype)
print("Number of dimensions:", a.ndim)
print("Size:", a.size)
```

```python
# create a 3X3 numpy array of all 9 
a = np.full((3, 3), 9)
print(a)
```

```python
# 4 cereate a 1d array of 10 evenly spaced values between 25 and 125
a = np.linspace(25,125,10)
print(a)
```

```python
# 5 convert a python list into numpy array 
list = [10,20,30,40,50]
a = np.array(list)
print(a)
```

```python
# 6 reverse a 1d numpy array 
a = np.array([1,2,3,4,5,6,7])
a = a[::-1]
print(a)
```

```python
# create a 4x4x3 array and extract value at second set,
#first row and last row

from numpy import reshape


a = np.arange(48).reshape(4,4,3)
print(a)
print(a[1,0,-1])
```

```python
# 8 create a 4X4 array and extract odd rows and even columns 
a = np.arange(1,17).reshape(4,4)
print("Original arrary:\n",a)

result = a[0::2,1::2]
print("Extracted odd rows and even columns:\n",result)
```

```python
# 9 slice first two row and firt two column ogf second set from a 4x4x3 array
a = np.arange(48).reshape(4,4,3)
result = a[1,:2,:2]
print(result)
```

```python
# 10 replace all odd number with -1 using for loop
a = np.array([[23,65,76,93],
              [71,82,13,24]])
for i in range(a.shape[0]):
    for j in range(a.shape[1]):
        if a[i,j] % 2 != 0:
            a[i,j] = -1
print(a)
```

```python
# 11. get indices of non-zero elements
a = np.array([1,0,2,0,3,0,4])
indices = np.nonzero(a)
print("Indices of non-zero elements:", indices)
```

```python
# 12 add two numpy array elments by element
a = np.array([1,2,3])
b = np.array([4,5,6])
result = a + b
print(result)
```

```python
# 13 multiply two numpy array elemnts by element
a = np.array([1,2,3])
b = np.array([4,5,6])
result = a * b
print(result)
```

```python
# 14 compute dot product of two numpy array
a = np.array([15,20,25])
b = np.array([10,40,37])
result = np.dot(a, b)
print(result)
```
