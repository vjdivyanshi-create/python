```python
import numpy as np
```

```python
a = np.array([[1,2,3],[4,5,6]])
b = np.array([[7,8,9],[10,11,12]])
res = np.concatenate((a,b), axis=1)
print(res)
```

```python
a = np.arange(9)
print(a)
result = np.split(a, 3)# here 3 is no. of elements in each splited array
print(result)

for i, a in enumerate(result):
    print("Splited array", i+1)
    print(a)
    
```

```python
a = np.array([[1, 2, 3], [4, 5, 6]])
print(a.shape)

# resizing the array
a.resize((3, 2))
print(a)
```

```python
a = np.array([[1, 2, 3], [4, 5, 6]])

res = np.append(a, [[7, 8,9]], axis=0)
print(res)
print("_________________")
res = np.insert(a, 1, [[10,11,12]], axis=0)
print(res)
```

```python
a = np.arange(12).reshape(3, 4)
print(a,"\n")
print(np.delete(a,6),"\n")
res = np.flip(a)
print(res)
```

```python
a = np.array([58,91,12,32,48,10,75,63,84,29])
res = np.sort(a)
print(res)

res = np.argmin(a)
print(res)

res = a[-1::-1]
print(res)
```

```python
a = np.array([[15, 27, 22], [19, 15, 5]])
b = np.array([[7, 8, 9], [4, 5, 2]])
res = a % b
print(res)
```
